**Theory**

Web scraping is the process of extracting data from websites automatically using programs or scripts. It allows users to collect large amounts of data from web pages without manually copying the information. Web scraping is widely used in data science for tasks such as data collection, price comparison, research, and market analysis.

In Python, web scraping is commonly performed using libraries like BeautifulSoup and Requests. The Requests library is used to send HTTP requests to a website and retrieve its HTML content, while BeautifulSoup is used to parse and extract specific data from that HTML.

**Q1: Scrape tabular data from a webpage**

Scraping tabular data means extracting data that is structured in rows and columns (HTML <table>). Using BeautifulSoup, we can locate the table using tags like <table>, <tr> (rows), and <td> or <th> (columns).

The process involves:

Sending a request to the webpage using Requests
Parsing the HTML content
Finding the table element
Looping through rows and extracting column data

This allows us to convert web tables into structured data for analysis.

csv module

**Q3: Handle missing or broken tags**

While scraping, some HTML elements may be missing or improperly structured. This can cause errors during extraction.
To handle this:


Use conditional checks like if tag: before accessing data


Use .get_text() with default values


Use try-except blocks to avoid crashes


Example approaches:


If a tag is missing, assign "N/A"


Skip incomplete rows


Handling such cases ensures that the scraping process is robust and does not fail due to minor issues in webpage structure.

**Q4: Use requests with headers**

Some websites block requests that do not come from a real browser. To avoid this, headers are added to the request using the Requests library.
Headers provide additional information such as:


User-Agent (browser type)


Accept-Language

```python
import requests
from bs4 import BeautifulSoup

url = "https://example.com"
response = requests.get(url)

soup = BeautifulSoup(response.text, "html.parser")

paragraphs = soup.find_all("p")

for p in paragraphs:
    print(p.text)
```

```python
import requests
from bs4 import BeautifulSoup

url = "https://example.com"
response = requests.get(url)

soup = BeautifulSoup(response.text, "html.parser")

links = soup.find_all("a")

for link in links:
    print(link.get("href"))
```
