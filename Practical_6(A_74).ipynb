**Theory**

Web scraping is the process of extracting data from websites automatically using programs or scripts. It allows users to collect large amounts of data from web pages without manually copying the information. Web scraping is widely used in data science for tasks such as data collection, price comparison, research, and market analysis.

In Python, web scraping is commonly performed using libraries like BeautifulSoup and Requests. The Requests library is used to send HTTP requests to a website and retrieve its HTML content, while BeautifulSoup is used to parse and extract specific data from that HTML.

**Q1: Scrape tabular data from a webpage**

Scraping tabular data means extracting data that is structured in rows and columns (HTML <table>). Using BeautifulSoup, we can locate the table using tags like <table>, <tr> (rows), and <td> or <th> (columns).

The process involves:

Sending a request to the webpage using Requests
Parsing the HTML content
Finding the table element
Looping through rows and extracting column data

This allows us to convert web tables into structured data for analysis.

```python
import requests
from bs4 import BeautifulSoup
```

```python
# Define the URL of the website you want to scrape
url = 'http://quotes.toscrape.com/' # A website specifically designed for scraping demonstrations

# Send an HTTP GET request to the URL
response = requests.get(url)

# Check if the request was successful (status code 200)
if response.status_code == 200:
    print(f"Successfully fetched content from {url}")
    # Get the HTML content of the page
    html_content = response.text
else:
    print(f"Failed to retrieve content. Status code: {response.status_code}")
    html_content = None
```

```python
if html_content:
    # Create a BeautifulSoup object
    soup = BeautifulSoup(html_content, 'html.parser')
    print("HTML content parsed successfully with BeautifulSoup.")
    # You can print a small part of the prettified HTML to see the structure
    # print(soup.prettify()[:1000]) # Print first 1000 characters of prettified HTML
else:
    soup = None
    print("Cannot parse HTML as content was not fetched.")
```

```python
if soup:
    # Extract the page title
    title = soup.title.text if soup.title else 'No Title Found'
    print(f"\nPage Title: {title}\n")

    print("Extracting quotes and authors:")
    # Find all div elements with class 'quote'
    quotes = soup.find_all('div', class_='quote')

    for quote in quotes:
        text = quote.find('span', class_='text').text
        author = quote.find('small', class_='author').text
        print(f"Quote: {text}\nAuthor: {author}\n")
else:
    print("Cannot extract data as BeautifulSoup object is not available.")
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

# Define the URL for a page with tabular data (e.g., Wikipedia)
tabular_url = 'https://en.wikipedia.org/wiki/List_of_countries_by_population_(United_Nations)'

# Add headers to mimic a browser request (Task 10)
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'
}

# Send an HTTP GET request to the URL with headers
tabular_response = requests.get(tabular_url, headers=headers)

# Check if the request was successful
if tabular_response.status_code == 200:
    print(f"Successfully fetched tabular content from {tabular_url}")
    tabular_html_content = tabular_response.text
    tabular_soup = BeautifulSoup(tabular_html_content, 'html.parser')
    print("Tabular HTML content parsed successfully with BeautifulSoup.")

    # Find the main table(s) on the page. Wikipedia tables often have the class 'wikitable'.
    # We'll try to find the first table with this class.
    table = tabular_soup.find('table', class_='wikitable')

    if table:
        # Extract table headers
        # Using .text.strip() to clean up header names
        headers = [th.text.strip() for th in table.find('tr').find_all('th')]
        print(f"Found table headers: {headers}")

        # Extract table rows (Task 6)
        table_data = []
        for row in table.find_all('tr')[1:]: # Skip the header row
            # Extract all 'td' and 'th' elements in the row for robustness
            row_data = [td.text.strip() for td in row.find_all(['td', 'th'])]
            table_data.append(row_data)

        # Handle missing or broken tags by ensuring consistent row lengths (Task 8)
        cleaned_table_data = []
        for row in table_data:
            if len(row) == len(headers):
                cleaned_table_data.append(row)
            elif len(row) > len(headers):
                # If a row has more columns than headers, truncate it
                cleaned_table_data.append(row[:len(headers)])
            else:
                # If a row has fewer columns, pad with empty strings
                cleaned_table_data.append(row + [''] * (len(headers) - len(row)))

        # Create a Pandas DataFrame
        df_countries = pd.DataFrame(cleaned_table_data, columns=headers)
        print("\nDataFrame created from tabular data:")
        display(df_countries.head())

        # Save the tabular data to a new CSV file (Task 7)
        output_tabular_csv_file = 'countries_by_population.csv'
        df_countries.to_csv(output_tabular_csv_file, index=False, encoding='utf-8')
        print(f"\nScraped tabular data successfully saved to '{output_tabular_csv_file}'")

    else:
        print("No table with class 'wikitable' found on the page.")

else:
    print(f"Failed to retrieve tabular content. Status code: {tabular_response.status_code}")
```
