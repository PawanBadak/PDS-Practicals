**Theory :-**


Correlation and regression are important statistical techniques used to study the relationship between two variables. Correlation measures the strength and direction of the relationship, while regression helps in predicting the value of one variable based on another. These methods are widely used in data analysis, economics, science, and machine learning.

**Q1. What is correlation? How is it different from regression?**

Correlation is a statistical method used to measure the strength and direction of relationship between two variables.
It tells us whether variables move together (positive), opposite (negative), or have no relation.

🔹 Difference between Correlation and Regression:


Correlation	Regression
Measures relationship	Predicts relationship
No dependent/independent variable	Has dependent (Y) and independent (X)
Value ranges from -1 to +1	Gives equation (Y = a + bX)
Symmetrical (X & Y same role)	Not symmetrical




**Q2. What is simple linear regression? Explain with example**



Simple Linear Regression is used to find a linear relationship between two variables and make predictions.

 Equation:
Y=a+bX
X → Independent variable
Y → Dependent variable
a → Intercept
b → Slope

 Example:
If study hours (X) increase, marks (Y) also increase.
Using regression, we can predict marks based on study hours.

**Q3. How do you calculate Pearson correlation in Python?**

```python
import numpy as np
from scipy.stats import pearsonr

X = [1, 2, 3, 4, 5]
Y = [2, 4, 5, 4, 5]

corr, _ = pearsonr(X, Y)

print("Pearson Correlation:", corr)
```

**Q4. What is multicollinearity? Does it affect simple regression?**

Multicollinearity occurs when independent variables are highly correlated with each other.

 Example:

Height in cm and height in inches (both give same info)
📌 Effect:
It affects multiple regression
It does NOT affect simple linear regression (because only one independent variable is used)

**Q5. A dataset shows strong correlation but poor prediction accuracy. Why?**

This can happen due to several reasons:

Non-linear relationship
→ Correlation checks linear relation only
Outliers in data
→ Few extreme values disturb prediction
Overfitting or underfitting
→ Model not properly trained
Correlation ≠ causation
→ Variables may be related but not useful for prediction
Limited data
→ Small dataset gives misleading results

```python
import pandas as pd

df = pd.read_csv("/content/train (1).csv")

df = df[['Age', 'Fare', 'Survived']]

df = df.dropna()

print(df.head())
```

```python
correlation = df.corr()

print("Correlation Matrix:\n")
print(correlation)
```

```python
import numpy as np
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt


X = df[['Age']]
y = df['Fare']

model = LinearRegression()
model.fit(X, y)

y_pred = model.predict(X)

print("Slope (m):", model.coef_[0])
print("Intercept (c):", model.intercept_)

plt.scatter(X, y)
plt.plot(X, y_pred)
plt.xlabel("Age")
plt.ylabel("Fare")
plt.title("Regression: Age vs Fare")
plt.show()
```

**Part-2**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

# Step 1: Load Dataset
df = pd.read_csv('/content/train (1).csv')

# Step 2: Select useful numeric columns
df = df[['Age', 'Fare', 'Pclass', 'SibSp', 'Parch', 'Survived']]

# Step 3: Handle missing values
df['Age'].fillna(df['Age'].mean(), inplace=True)

# Step 4: Define X and y
X = df[['Age', 'Fare', 'Pclass', 'SibSp', 'Parch']]  # Features
y = df['Survived']  # Target

# Step 5: Train Model
model = LinearRegression()
model.fit(X, y)

# Step 6: Model Parameters
print("Coefficients:", model.coef_)
print("Intercept:", model.intercept_)

# Step 7: Predictions
y_pred = model.predict(X)

# Step 8: Evaluation
mse = mean_squared_error(y, y_pred)
r2 = r2_score(y, y_pred)

print(f"MSE: {mse:.4f}")
print(f"R2 Score: {r2:.4f}")

# Step 9: Visualization (Actual vs Predicted)
plt.figure(figsize=(8,5))
plt.scatter(y, y_pred)
plt.xlabel("Actual Survived")
plt.ylabel("Predicted Survived")
plt.title("Actual vs Predicted (Linear Regression)")
plt.grid()
plt.show()
```
