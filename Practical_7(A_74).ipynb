**Theory**


Probability distributions are an important part of statistics and data analysis, as they describe how the values of a random variable are spread. A random variable is a quantity whose value depends on chance, such as the number of heads in coin tosses or marks scored by students. In Python, probability distributions can be easily implemented and analyzed using libraries like numpy, scipy.stats, and matplotlib. These tools help in calculating probabilities, generating random data, and visualizing results through graphs, making it easier to understand patterns and behavior of data.

## Answers to Questions

### 1. What is the difference between Binomial, Poisson, and Normal distributions?

*   **Binomial Distribution:** Describes the number of successes in a fixed number of independent Bernoulli trials (experiments with only two outcomes, success or failure), each with the same probability of success. It is a discrete distribution.
    *   **Parameters:** `n` (number of trials) and `p` (probability of success).
    *   **Example:** The number of heads when flipping a coin 10 times.

*   **Poisson Distribution:** Describes the number of events occurring in a fixed interval of time or space, given a constant average rate of occurrence and independence of events. It is a discrete distribution.
    *   **Parameter:** `λ` (lambda), the average rate of events.
    *   **Example:** The number of calls received by a call center in an hour.

*   **Normal Distribution (Gaussian Distribution):** A continuous probability distribution that is symmetric about its mean, showing that data near the mean are more frequent in occurrence than data far from the mean. It is characterized by its bell-shaped curve.
    *   **Parameters:** `μ` (mean) and `σ` (standard deviation).
    *   **Example:** Heights of adult males in a population.

### 2. What is the real-world significance of the Poisson distribution?

The Poisson distribution is highly significant in real-world scenarios for modeling the occurrence of rare events over a fixed period or in a defined space. Its applications include:

*   **Quality Control:** Number of defects per unit of product.
*   **Queuing Theory:** Number of customers arriving at a service desk in a given interval.
*   **Epidemiology:** Number of disease cases in a certain region over a specific time.
*   **Insurance:** Number of claims filed per day.
*   **Biology:** Number of mutations in a DNA strand.
*   **Telecommunications:** Number of incoming calls to a call center.

It helps in understanding and predicting the frequency of events, which is crucial for resource allocation, risk management, and operational planning.

### 3. What is the Central Limit Theorem and how does it relate to Normal distribution?

**Central Limit Theorem (CLT):** States that if you take sufficiently large random samples from any population with a finite mean (μ) and standard deviation (σ), the distribution of the sample means will be approximately normally distributed, regardless of the population's original distribution. As the sample size increases, the approximation gets better.

**Relation to Normal Distribution:** The CLT is fundamental to why the Normal distribution is so prevalent in statistics. It explains why many natural phenomena, when observed through repeated sampling, tend to exhibit a normal distribution. Even if individual data points don't follow a normal distribution, the means of samples taken from that data will tend to be normally distributed. This allows statisticians to use normal distribution properties for hypothesis testing, confidence interval construction, and other inferential statistics, even when the underlying population distribution is unknown or non-normal.

### 4. Why is Normal distribution widely used in machine learning and statistics?

The Normal distribution is widely used due to several key properties and theoretical underpinnings:

*   **Central Limit Theorem:** As explained above, it implies that sample means from almost any distribution will be normally distributed, making it a powerful tool for inference.
*   **Mathematical Tractability:** Its probability density function is well-defined and has many convenient mathematical properties that simplify calculations and theoretical derivations.
*   **Natural Phenomena:** Many real-world phenomena (e.g., height, weight, measurement errors, test scores) naturally approximate a normal distribution.
*   **Parameter Estimation:** For normally distributed data, the sample mean and variance are efficient estimators of the population mean and variance.
*   **Basis for Statistical Tests:** Many classical statistical tests (e.g., t-tests, ANOVA) assume normality of the data or residuals.
*   **Machine Learning Algorithms:** Many algorithms assume or perform better with normally distributed data. For example, linear regression assumes normally distributed residuals, and some classification algorithms like Linear Discriminant Analysis assume normally distributed classes.

### 5. What is skewness in distributions and which distributions exhibit it?

**Skewness:** Is a measure of the asymmetry of the probability distribution of a real-valued random variable about its mean. A distribution is skewed if one of its tails is longer than the other.

*   **Positive Skew (Right-skewed):** The tail on the right side of the distribution is longer or fatter than the left side. The mean is typically greater than the median, which is greater than the mode.
*   **Negative Skew (Left-skewed):** The tail on the left side of the distribution is longer or fatter than the right side. The mean is typically less than the median, which is less than the mode.

**Distributions that exhibit skewness:**

*   **Poisson Distribution:** Can be positively skewed, especially for smaller values of λ. As λ increases, the Poisson distribution becomes more symmetric and approaches a normal distribution.
*   **Binomial Distribution:** Can be skewed if the probability of success (p) is not 0.5. If p < 0.5, it's positively skewed. If p > 0.5, it's negatively skewed. If p = 0.5, it's symmetric.
*   **Exponential Distribution:** Always positively skewed.
*   **Log-Normal Distribution:** Always positively skewed.
*   **Chi-squared Distribution:** Always positively skewed.
*   **Weibull Distribution:** Can be positively or negatively skewed depending on its parameters.

```python
# Import libraries
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom, poisson, norm

# -------------------------------
# 1. BINOMIAL DISTRIBUTION
# -------------------------------
print("Binomial Distribution")

n = 10      # number of trials
p = 0.5     # probability of success
x = np.arange(0, n+1)

binomial_pmf = binom.pmf(x, n, p)

plt.figure(figsize=(6,4))
plt.bar(x, binomial_pmf)
plt.title("Binomial Distribution (n=10, p=0.5)")
plt.xlabel("Number of Successes")
plt.ylabel("Probability")
plt.tight_layout()
plt.show()


# -------------------------------
# 2. POISSON DISTRIBUTION
# -------------------------------
print("Poisson Distribution")

lam = 3   # average rate
x = np.arange(0, 10)

poisson_pmf = poisson.pmf(x, lam)

plt.figure(figsize=(6,4))
plt.bar(x, poisson_pmf)
plt.title("Poisson Distribution (λ=3)")
plt.xlabel("Number of Events")
plt.ylabel("Probability")
plt.tight_layout()
plt.show()


# -------------------------------
# 3. NORMAL DISTRIBUTION
# -------------------------------
print("Normal Distribution")

mean = 0
std_dev = 1

x = np.linspace(-4, 4, 100)
normal_pdf = norm.pdf(x, mean, std_dev)

plt.figure(figsize=(6,4))
plt.plot(x, normal_pdf)
plt.title("Normal Distribution (mean=0, std=1)")
plt.xlabel("Values")
plt.ylabel("Density")
plt.tight_layout()
plt.show()
```
