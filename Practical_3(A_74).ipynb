Q2. Importance of EDA in Pandas

Exploratory Data Analysis (EDA) is a crucial step in any data analysis project, and Pandas provides excellent tools for it. Here's why EDA is important:

Understand Data Structure: EDA helps you understand the format, structure, and types of data (e.g., numerical, categorical, dates) within your Pandas DataFrame.
Identify Missing Values: You can easily pinpoint missing data points and decide on strategies to handle them (e.g., imputation, deletion).
Detect Outliers: EDA helps in identifying unusual data points that might skew your analysis or model training, which can then be investigated or treated.
Discover Patterns and Relationships: By visualizing and summarizing data, you can uncover hidden patterns, correlations between variables, and trends that might not be obvious at first glance.


Q3. Difference between Series, DataFrame, and Panel



Let's break down the differences between Series, DataFrame, and Panel in Pandas:

Series:

1-dimensional labeled array: Think of it like a single column in a spreadsheet or a Python list with an added index.
Homogeneous data: All elements in a Series must be of the same data type (e.g., all integers, all strings, all floats).
Structure: It has a data component and an index component.
Analogy: A single column of data with row labels.
Use cases: Storing a single variable's data, time series data, or as a component of a DataFrame.
DataFrame:

2-dimensional labeled data structure: It's the most commonly used Pandas object, resembling a table or spreadsheet.
Heterogeneous data: Each column can have a different data type, but all elements within a single column must be of the same type.
Structure: Composed of multiple Series, where each Series represents a column. It has both row (index) and column labels.
Analogy: A spreadsheet with rows and columns, where each column can hold different types of data.
Use cases: Storing tabular data, performing data cleaning, manipulation, and analysis.

```python
import pandas as pd

df = pd.read_csv("/content/train (1).csv")
df.head()
```

```python
age_column = df["Age"]
print(age_column.head())
```

```python
df["Source"] = "Titanic Dataset"
print(df[["Name", "Source"]].head())
```

```python
sorted_df = df.sort_values(by="Fare", ascending=False)
print(sorted_df[["Name", "Fare"]].head())
```

```python
df_renamed = df.rename(columns={
    "Pclass": "Passenger_Class",
    "SibSp": "Siblings_Spouses",
    "Parch": "Parents_Children"
})
print(df_renamed.columns.tolist())
```

```python
print(df.describe())
```

```python
grouped = df.groupby("Pclass")[["Fare", "Age"]].mean()
print(grouped)
```

```python
extra_info = pd.DataFrame({
    "PassengerId": [1, 2, 3, 4, 5],
    "Nationality": ["British", "American", "Irish", "French", "German"]
})

merged_df = pd.merge(df, extra_info, on="PassengerId", how="left")
print(merged_df[["PassengerId", "Name", "Nationality"]].head())
```

```python
df_dropped_cols = df.drop(columns=["Cabin", "Ticket", "Source"])
print("After dropping columns:", df_dropped_cols.shape)
```

```python
df_dropped_rows = df_dropped_cols.dropna(subset=["Age"])
print("After dropping rows with missing Age:", df_dropped_rows.shape)
```

```python
def age_group(age):
    if age < 18:
        return "Child"
    elif age < 60:
        return "Adult"
    else:
        return "Senior"

df["Age_Group"] = df["Age"].apply(age_group)
print(df[["Name", "Age", "Age_Group"]].head(10))
```
