**Theory**

Data visualization is an important part of data analysis that involves representing data in graphical form such as charts, graphs, and plots. It helps in simplifying complex data and makes it easier to understand patterns, trends, and relationships. In data analysis, visualization plays a key role in decision-making by providing clear insights from large datasets.

In Exploratory Data Analysis (EDA), data visualization is very useful as it helps in understanding the structure of the dataset. By using graphs like histograms, bar charts, and scatter plots, we can identify the distribution of data, detect outliers, and observe relationships between variables. Visualization allows analysts to quickly explore the data and form hypotheses for further analysis.

**Q3) Purpose of the pyplot Module in Matplotlib.**

**Q4) Process of Creating a Basic Linear Plot**

A linear plot (line graph) shows the relationship between two variables.

Steps:

Import library

import matplotlib.pyplot as plt

Define data

x = [1, 2, 3]
y = [4, 5, 6]

Create plot

plt.plot(x, y)

Add labels and title

plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Basic Line Plot")

Display graph

plt.show()


**Q5) Importance of Customization in Data Visualization & Options in Matplotlib**

Why Customization is Important:
Makes graphs more clear and readable
Improves presentation quality
Highlights important information
Makes data visually appealing
Customization Options in Matplotlib:
Titles: plt.title()
Labels: plt.xlabel(), plt.ylabel()
Colors: color='red'
Line styles: linestyle='--'
Markers: marker='o'
Legends: plt.legend()
Grid: plt.grid()
Figure size: plt.figure(figsize=(6,4))

👉 Example:

plt.plot(x, y, color='green', linestyle='--', marker='o')
plt.title("Customized Plot")
plt.grid()

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv("https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv")
df.head()
df["Age"] = df["Age"].fillna(df["Age"].mean())
df.drop("Cabin", axis=1, inplace=True)
plt.plot(df["Age"])
plt.title("Age Line Plot")
plt.show()
plt.scatter(df["Age"], df["Fare"])
plt.xlabel("Age")
plt.ylabel("Fare")
plt.title("Age vs Fare")
plt.show()
df["Pclass"].value_counts().plot(kind='bar')
plt.title("Passenger Class Count")
plt.show()
plt.hist(df["Age"])
plt.title("Age Distribution")
plt.show()
sns.boxplot(x=df["Age"])
plt.title("Age Box Plot")
plt.show()
sns.pairplot(df)
```
